# Quantum vs Classical Portfolio Optimization — Exploration

Interactive companion to `src/main.py`. Use this notebook to explore the efficient frontier, inspect QAOA convergence, and visualize weight allocations.

```python
import sys
sys.path.append('..')

import yaml
from src.data_loader import fetch_prices, compute_returns
from src.utils import annualized_return, annualized_covariance
from src.classical_opt import max_sharpe_portfolio
from src.quantum_opt import qaoa_portfolio

with open('../config/stocks.yaml') as f:
    config = yaml.safe_load(f)

prices = fetch_prices(config['tickers'], config['start_date'], config['end_date'])
returns = compute_returns(prices)
mean_returns = annualized_return(returns)
cov_matrix = annualized_covariance(returns)
prices.tail()
```

```python
classical_result = max_sharpe_portfolio(mean_returns, cov_matrix)
print(classical_result.summary())
```

```python
qaoa_cfg = config['qaoa']
quantum_result = qaoa_portfolio(
    mean_returns, cov_matrix,
    budget=qaoa_cfg['budget'],
    risk_factor=qaoa_cfg['risk_factor'],
    reps=qaoa_cfg['reps'],
    max_iter=qaoa_cfg['max_iter'],
    backend='simulator',
)
print(quantum_result.summary())
```
